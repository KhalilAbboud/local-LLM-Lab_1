```python
import torch.nn as nn

model = nn.Sequential(
    nn.Linear(3, 4),
    nn.ReLU(),

    nn.Linear(4, 4),
    nn.ReLU(),

    nn.Linear(4, 1)
)

print(model)
total_params = sum(p.numel() for p in model.parameters())
print("Nombre total de paramètres :", total_params)

for name, param in model.named_parameters():
    print(name, param.shape)
```
